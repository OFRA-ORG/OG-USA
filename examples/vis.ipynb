```python
import os, glob
import pandas as pd
```

```python
pd.options.display.float_format = "{:,.3f}".format
```

```python
df_cbo = pd.read_csv('cbo.csv')
df_cbo['Revenues (trillions)'] = df_cbo['Revenues (pct of GDP)'] * df_cbo['Nominal GDP (trillions)'] / 100
df_cbo['Spending (trillions)'] = df_cbo['Spending (pct of GDP)'] * df_cbo['Nominal GDP (trillions)'] / 100
```

```python
def add_entry(idx, folder_name, df_cbo, results_dict):
    file_name = folder_name + '/plots_and_tables/macro_time_series_output.csv'
    df = pd.read_csv(file_name)
    df = df[df['Year']<=2035]
    pct_diff_rev = (df['Total tax revenue ($REV_t$): Reform'] - df['Total tax revenue ($REV_t$): Baseline'])/df['Total tax revenue ($REV_t$): Baseline']

    df_cbo_i = df_cbo.copy()
    dollar_diff_rev = pct_diff_rev * df_cbo_i['Revenues (trillions)']

    results_dict[idx] = [folder_name] + (pct_diff_rev * df_cbo_i['Revenues (trillions)']).to_list()
```

```python
# Initializing an empty dictionary 
results_dict = {} 
```

### Based sim

```python
model = 'TCJA_TMD'
frisch = '0.4'
zeta_D = '0.4'
g_y_annual = '0.02'
tG1 = '20'
folder_name = model+"##frisch##"+frisch+"##zeta_D##"+zeta_D+"##g_y_annual##"+g_y_annual+"##tG1##"+tG1
```

```python
add_entry(0, folder_name, df_cbo, results_dict)
```
